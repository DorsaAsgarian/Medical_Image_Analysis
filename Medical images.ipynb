# <center>What is DICOM<center>

DICOM is the international standard for medical image and represented as .dcm. So, it's the output format of most medical imaging devices used in hospitals and clinical car setting, such as MRI, PET, CT, Xray, and ultrasound. 

There are different types of medical imaging and they are different from each other. For instance x-ray is a 2-dimensional and monochromatic image while ultrasound is a 3-dimension image or even a video. Also, these medical images produced by different machines which are from different companies. So, the diversity makes a situation in which there is need for a standard so that radiologists can see all medical images in their computer. 
<br>
DICOM files have a *print* function so the radiologists can print them.
<br><br>
**Differences between DICOM and JPEG?** 
<br>
A JPEG file is `one` monochromatic image. It does not contain metadata. 
<br>
DICOM could have several images. Also it contains metadata. It differs from other image formats because it groups information into datasets. DICOM file consist of header and image data collectively in one file.

There are some online tools to convert DICOM files to JPG. I used <a href="https://convertio.co/dcm-jpg/">convertio</a> website to convert a DICOM x-ray to JPG.
The output of DICOM file is in X-ray section.

### JPEG sample 


X-ray images are 2-dimension grayscale (high resolution) images which means some pixels are darker and some pixels are brighter. In computer vision terms it means that pixels values ranging from 0 to 255. <br>
0: completely dark pixels.
<br>
255: completely white pixels. 

<img src="https://miro.medium.com/max/1024/1*5wCkABhpgGisONRV1ZxGoA.webp">

X-rays only displayes 5 densities:
- Dark: Locations on the body which are filled with *air*. These areas appear black.

- Dark gray: Subcutaneous tissues(بافت زیرجلدی) or *fat*.

- Light gray: *soft tissues* like the heart and blood vessels.

- Off white: *bones* such as the ribs.

- bright white: presence of *metallic objects*.

**Interpreting an X-ray image:**<br> 
The physician looks at the borders between the different densities.
In the image below, the ribs appear off-white because they are dense tissues, but since the lungs are filled with air, they appear dark.
Below the lung is the hemidiaphragm, which is a soft tissue and appears light grey. According to these different densities, there are demarcated lines in an X-ray images.
Now, if something were to happen in the lungs, such as pneumonia, then the air dense lungs will change into waters-dense-lungs. This would cause the demarcation lines to fade.

### Concept of windowing

In the windowing process, the grayscale of a particular image can be adjusted. <br>
Windowing uses two values for getting a right window:
- window width: the range of the grayscale that can be displayed. This one controls contrast
- window level: the centre of grayscale range. This one controls brightness.

Windowing also known as grey-level mapping, contrast stretching, histogram modification or contrast enhancement.

### Physics of MRI

MRI is based on the magnetization properties of atomic nuclie. 
<br>
In tissue, there are randomly oriented water nuclie. In MRI there is a powerful magnetic field that employed to align the protons of these nuclei.
<br> 
There is also an exteranl Radio Frequency (RF) energy. The water nuclei return to their resting position through various relaxation processes.This results in emiting RF energy.
<br>
After a certain period following the initial RF, the emitted signals are measured using fourier transformation. Fourier transformation convert the frequency information contained in the signal from each location in the imaged plane to corresponding intensity levels, which are then displayed as shades of grey in a matrix of pixels. 
<br>
By varying the sequence of RF pulses, different types of images are created. 
<br>
<font color="blue">Repetition Time (TR) </font> is the amount of time between successive pulse sequence applied to the same slice.
<br>
<font color="blue">Time to Echo (TE) </font> is the time between the delivery of the RF pulse and the receipt of the eacho signal. 

### MRI image sequences

The most common MRI sequences are T1-weighted and T2-weighted scans.

**T1-Weighted images** are produced by using short TE and TR times. 
<br>
**T2-Weighted images** are produced by using longer TE and TR times. 

There is another sequence called **Fluid Attenuated Inversion Recovery (FLAIR)**. The FLAIR sequence is similar to a T2-weighted image, expect that the TE and TR are very long. 

Here, is an image of comparision of T1, T2 and FLAIR:

**Structural and functional MRI**<br>
Structural MRI is a generic term for any MRI scan intended to image the structure of the body. These are sometimes called anatomical scans. <br>
This is in contrast to functional MRI (FMRI, which are scans designed to measure some aspect of physiological function like neural activation and brain activity).

```python
# reading DICOM files
def readfile ():
    # path of dicome folder, this folder contains all 109 subject's folder
    dicom_path = os.path.join(os.getcwd(), "dicom-flair")
    
    # enter to the first subject's folder, this folder has 29 .dcm files
    first_file = os.path.join(dicom_path, os.listdir(dicom_path)[0])
    
    # for every .dcm file in folder
    for file in os.listdir(first_file):
        # full path and file name of the .dcm file
        file_name = os.path.join(first_file, file)
        # add it to the dataset
        read_dicom = dcmread(file_name)
        
    return read_dicom
```

**Header analysis:**
- Size of header: stores the size of the header. It must be 348 for a nifti or analyze format.

- Dim info: stores the freauency encoding direction (1, 2 or 3), the phase encoding direction (1, 2, or 3), and the direction in which the volume was sliced during the acquisition (1, 2 or 3). 

- dim[ ]: contains the size of the image array. The dim[0] contains the number of dimensions (1-7). The dimensions 1,2 and 3 are assumed to refer to space (x,y, z), the 4th dimension is assumed to refer to time, ans the remaining dimensions (5,6, and 7) can be anything else. The value dim [i] is a positive integer representing the length of the i-th dimension. 

- datatype: indicates the type of the data stored. 

- bitpix: holds the information of the number of bits per voxel. The value must match the type determined by `datatype` above.

- slice_codes are useful to store information about the timing acquisition. For example 0 means slice order unknown, 1 means sequentioal and increasing, 2 means sequential and decreasing, etc. 

    - slice_start and slice_end inform respectively which are the first and last slices.
    - slice_duration indicates the amount of time needed to asquire a single slice. 
You can see the example of slice_codes below:

<img src="https://s3.us-east-2.amazonaws.com/brainder/2012/nifti/slicetimingcodes_90dpi.png">

- vox_offset: indicates, for sn=ingle files (.nii), the byte offset before the imaging data starts for compatibility with older software. This should be zero if no information other than image data itsel is to be stored in the .img, but it can also be larger than zero, allowing extra information to be prepended into the .img (such as DICOM header). 

- scl_slope and scl inter: the values stored in each voxel can be linearly scaled to different uints. These two fields define a slope and an intercept for a linear function. The dat  scaling feature allows the storage in a wider range than what would be allowed bt the datatype. 

- cal_max and cal_min: for files that store scalar (non-vector) data, these fileds determine the intended display range when the image is opened. Voxel values equal or below `cal-min` should be shown with the smallest color in the colorscale (typically black in a gray-scale visualisation) , and values equal or above `cal_max` should be shown with the largest color in the colorscale (typically white).

- descrip: can contains any text with up to 80 characters. 

- xyzt_units: both spatial and temporal measurement units, used for the dimensions dim[1] to dim[4] are ecnoded in this field. The bits 1-3 are used to sotre the spatial dimensions, the bits 4-6 are for temporal dimensions, and the bit 6 and 7 are not used. A temporal offset can be specified in the field `toffest`. 

- aux_files: A supplementary file, containing extra_information, can be specified in this field.

- qform_code and sform_code: orientation information.

#### What is Ultrasound?

Ultrasound waves are produced by a transducer, which can both emit ultrasound waves, as well as detect the ultrasound echoes reflected back.<br>

<img src="https://www.nibib.nih.gov/sites/default/files/inline-images/Ultrasound_Terese%20Winslow.jpg">

The active elements in ultrasound transducers are made of special ceramic crystal materials called piezoelectrics. These materials are able to produce sound waves when an electric field is applied to them, but can also work in reverse, producing an electric field when a sound wave hits them. 

There are two main variety of ultrasound: 
- B-mode: the reflected sound waves create simple and still image of the anatomy. These still images are balck and white, making them 2-dimensional.
- Doppler: the distortion of the sound waves due to movement in the body is used to show the flow of fluids, such as blood through the veins. These movements are color-coded, making doppler ultrasound scans 3-dimensional. 

In both B-Mode and Doppler, still images can be taken in quick succession to create videos.
